### Ejercicio B1: Escribir una lista en un fichero

Escribe una lista de 5 nombres en un fichero de texto `nombres.txt`, uno por linea, usando `with open(...) as f:`.

```python
# Tu codigo aqui
nombres = ["jaime", "Adrian", "Jose"]
with open('salida.txt', 'a', encoding='utf-8') as f:
    for i in nombres:
        f.write(i + "\n")
```

### Ejercicio I1: Guardar alumnos en CSV

Guarda una lista de diccionarios de alumnos (`nombre`, `nota`) en un fichero `alumnos.csv` usando el modulo `csv` con `DictWriter`, y luego vuelve a leerlo con `DictReader` mostrando cada fila.

```python
# Tu codigo aqui
import csv
alumnos = {"Jaime" : 6 , "Adrian": 7 , "Jose" : 8}
with open('datos.csv', 'w', newline='', encoding='utf-8') as f:
    escritor = csv.writer(f)
    escritor.writerow(["Nombre", "Nota"])
    for nombre, nota in alumnos.items():
        escritor.writerow([nombre, nota])

    
with open('datos.csv', 'r', encoding='utf-8') as f:
    for fila in csv.reader(f):
        print(fila)
```

### Ejercicio I2: Configuracion en JSON

Crea un diccionario de configuracion (`idioma`, `tema`, `notificaciones`), guardalo en `config.json` con `json.dump`, y vuelve a cargarlo con `json.load` para comprobar que coincide.

```python
# Tu codigo aqui
import json
configuracion = {"idioma": "español", "tema": "oscuro", "notificaciones": True}
json_string = json.dumps(configuracion, indent=2, ensure_ascii=False)


with open('config.json', 'w', encoding='utf-8') as f:
    json.dump(configuracion, f, indent=2, ensure_ascii=False)
    

datos_leidos = json.loads(json_string)
with open('config.json', 'r', encoding='utf-8') as f:
    datos_desde_archivo = json.load(f)
print(datos_desde_archivo)
```

### Ejercicio A1: Informe con marca temporal

Usando `os.path` para construir la ruta y `datetime` para la marca temporal, escribe una funcion que genere un fichero `informe_<fecha>.txt` con un resumen de ventas (diccionario producto->importe) y el total.

```python
# Tu codigo aqui
import datetime
import os.path

fecha=(datetime.date.today())
ruta = os.path.join("informe_"+str(fecha)+".txt")

ventas = {"Monitor": 500,"Raton": 50,"Teclado": 70}

with open(ruta, 'w', encoding='utf-8') as f:
    for producto, importe in ventas.items():
        f.write(f"{producto} : {importe}\n")

```
